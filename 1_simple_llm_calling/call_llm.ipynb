```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model="gemma-4-31b-it")
response = llm.invoke("How many teeth does human have?")
print(response.text)
```

```python
response = llm.invoke([
    ["system", "You are an AI assistant that only answers in one or two lines"],
    ["human", "How many teeth does human have?"]
])

print(response.text)
```

```python
prompt = "Write a single line to impress the interviewer, just answer in one line"

llm = ChatGoogleGenerativeAI(model="gemma-4-31b-it", temperature=0) #with increase in temperature, the creativity increases
response = llm.invoke(prompt)
print(response.text)
```

```python
llm = ChatGoogleGenerativeAI(model="gemma-4-31b-it", temperature=2) #with increase in temperature, the creativity increases
response = llm.invoke(prompt)
print(response.text)
```

**CALL LLM USING GROQ MODEL**

```python
from langchain_groq import ChatGroq

llm = ChatGroq(model="qwen/qwen3.6-27b", reasoning_format="parsed")
messages = [
    ("system", "You are an AI assistant that only answers in one or two lines"),
    ("human", "How many teeth does human have?")
]

ai_msg=llm.invoke(messages)
print(ai_msg.content)
```
