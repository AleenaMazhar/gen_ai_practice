```python
import chromadb

client = chromadb.Client()
```

```python
collection = client.create_collection("news_articles")
```

```python
collection.add(
    ids = ["id1", "id2", "id3", "id4"],
    documents = [
        "Apple is a electronics product company",
        "Tesla is a car making company",
        "SpaceX got NASA contract",
        "Apples are high in fiber"
    ]
)
```

```python
data = collection.get(
    include=["documents", "embeddings", "metadatas"]
)

for i, emb in enumerate(data["embeddings"]):
    print(f"\nDocument {i+1}")
    print("Text:", data["documents"][i])
    print("Embedding Length:", len(emb))
    print("First 10 values:", emb[:10])
```

```python
results = collection.query(
    query_texts = ["This is a query related to Elon Must"],
    n_results = 2
)

print(results)
```
